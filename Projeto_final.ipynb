```python
livros=[
  {"titulo": "Crime e Castigo",
        "autor": "Fiódor Dostoiévski ",
        "ano": "1866",
        "disponivel": True }
]
```

```python
def cadastrar_livro():
    titulo = input("Digite o título do livro: ")
    autor = input("Digite o nome do autor: ")
    ano = input("Digite o ano de publicação: ")

    livro = {"titulo": titulo,
        "autor": autor,
        "ano": ano,
        "disponivel": True }

    livros.append(livro)
    print("Livro cadastrado com sucesso!")

def listar_livros():
    if len(livros) == 0:
        print("Nenhum livro cadastrado.")
    else:
        for livro in livros:
            print("--------------------")
            print("Título:", livro["titulo"])
            print("Autor:", livro["autor"])
            print("Ano:", livro["ano"])
            print("Disponível:", livro["disponivel"])


def buscar_livro():
    busca = input("Digite o título ou autor do livro: ")

    encontrado = False

    for livro in livros:
        if busca.lower() in livro["titulo"].lower() or busca.lower() in livro["autor"].lower():
            print("--------------------")
            print("Título:", livro["titulo"])
            print("Autor:", livro["autor"])
            print("Ano:", livro["ano"])
            print("Disponível:", livro["disponivel"])
            encontrado = True

    if encontrado == False:
        print("Livro não encontrado.")


def emprestar_livro():
    titulo = input("Digite o título do livro que deseja emprestar: ")

    encontrado = False

    for livro in livros:
        if titulo.lower() == livro["titulo"].lower():
            encontrado = True

            if livro["disponivel"] == True:
                livro["disponivel"] = False
                print("Empréstimo realizado com sucesso!")
            else:
                print("Este livro já está emprestado.")

    if encontrado == False:
        print("Livro não encontrado.")




def devolver_livro():
    titulo = input("Digite o título do livro que deseja devolver: ")

    encontrado = False

    for livro in livros:
        if titulo.lower() == livro["titulo"].lower():
            encontrado = True

            if livro["disponivel"] == False:
                livro["disponivel"] = True
                print("Devolução realizada com sucesso!")
            else:
                print("Este livro já está disponível.")

    if encontrado == False:
        print("Livro não encontrado.")


while True:
    print("\n===== BIBLIOTECA =====")
    print("1 - Cadastrar livro")
    print("2 - Listar livros")
    print("3 - Buscar livro")
    print("4 - Emprestar livro")
    print("5 - Devolver livro")
    print("6 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        cadastrar_livro()
    elif opcao == "2":
        listar_livros()
    elif opcao == "3":
        buscar_livro()
    elif opcao == "4":
        emprestar_livro()
    elif opcao == "5":
        devolver_livro()
    elif opcao == "6":
        print("Programa encerrado!")
        break
    else:
        print("Opção inválida! Tente novamente.")
```
